```python
import pandas as pd
import numpy as np
```

loading Data

```python
movies = pd.read_csv("./tmdb_5000_movies.csv")
credits = pd.read_csv("./tmdb_5000_credits.csv")

credits.head()
```

```python
movies.head(1)
```

```python
movies = movies.merge(credits,on='title')
```

```python
movies.shape
movies.info()
```

```python
# genres
# id
# Keywords
# title
# overview
# cast
# crew
movies = movies[['genres','movie_id','keywords','title','overview','cast','crew']]
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace = True)
```

```python
# movies.isnull().sum()
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
import ast

def convertToList(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
        
    return L
```

```python
# convertToList('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')

movies['genres'] = movies['genres'].apply(convertToList)
```

```python
movies.head(1)
```

```python
movies['keywords'] = movies['keywords'].apply(convertToList)

```

```python
movies.head(1)
```

```python
movies['cast'][0]
```

```python
def convertToListofActor(obj):
    L = []
    count = 0
    for i in ast.literal_eval(obj):
        if count != 3:
            L.append(i['name'])
            count+=1
        else:
            break    
        
    return L
```

movies['cast'] = movies['cast'].apply(convertToListofActor)

```python
movies['crew'][0]
```

```python

def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
        
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head(1)
```

```python
movies['overview'][0]
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
```

```python
movies['keywords'] = movies['keywords'].apply(lambda x :[i.replace(" ","") for i in x])
movies['genres'] = movies['genres'].apply(lambda x :[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x :[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x :[i.replace(" ","") for i in x])
```

```python
movies.head(1)
```

```python
movies['tags'] = movies['genres'] + movies['keywords'] + movies['overview'] + movies['cast'] + movies['crew']
```

```python
movies.head(1)
```

```python
new_df = movies[['movie_id','title','tags']]
```

```python
new_df.head(1)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df.head(1)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
new_df['tags'][0]
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 1000,stop_words = 'english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
def recomended(movies):
    movie_index = new_df[new_df['title'] == movies].index[0]
    distance = similarity[movie_index]
    movie_list = sorted(enumerate(distance),reverse=True,key = lambda x:x[1])[1:6]
    
    for i in movie_list:
        print(new_df.iloc[i[0]].title)
```

```python
recomended('Avatar')
```
