</h1>Colabrative Filtering Based Recommender System

```python
x = ratings_with_name.groupby('User-ID').count()['Book-Rating']>200
inteligent_user = x[x].index
```

```python
filted_rating = ratings_with_name[ratings_with_name['User-ID'].isin(inteligent_user)]
```

```python
y = filted_rating.groupby('Book-Title').count()['Book-Rating']>=50
famous_bookes = y[y].index
```

```python
final_ratings = filted_rating[filted_rating['Book-Title'].isin(famous_bookes)]
```

```python
pt = final_ratings.pivot_table(index='Book-Title',columns='User-ID',values = 'Book-Rating')
```

```python
pt.fillna(0,inplace=True)
```

```python
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_score = cosine_similarity(pt)
similarity_score
```

```python
def recommandation(name):
    # index fetch from book
    index = np.where(pt.index == name)[0][0]
    similar_items = sorted(list(enumerate(similarity_score[index])),key = lambda x :x[1],reverse=True)[1:6]
    
    for i in similar_items:
        print(pt.index[i[0]])
    
    # return suggestion
```

```python
recommandation('Message in a Bottle')
```
